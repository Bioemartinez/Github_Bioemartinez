```python
# Para realizar cálculos individuales repitiendo la fórmula existe el siguiente código
```

```python
#Como ejemplo, usemos el cálculo de los componentes de la corriente en la dirección Este, u, y Norte, v, tomando la rapidez (magnitud del vector de velocidad) y el angulo (ángulo en la convención geográfica, en grados, donde los valores aumentan en el sentido del reloj de 0, Norte, a 360):
```

```python
import math
```

```python
# En km/h y grados
rapidez1, angulo1 = 1.0, 30.0
```

```python
u = rapidez1 * math.sin(math.radians(angulo1))
v = rapidez1 * math.cos(math.radians(angulo1))

u, v
```

```python
# Ahora es necesario hacer la transformación de otro vector de velocidad
```

```python
rapidez2, angulo2 = 2.0, 90.0
```

```python
# Si tenemos que repetir las fórmulas ...
```

```python
u = rapidez2 * math.sin(math.radians(angulo2))
v = rapidez2 * math.cos(math.radians(angulo2))

u, v
```

```python
# Y si tenemos otro vector, también debemos repetir las fórmulas una tercera vez!
```

```python
## Sin embargo, podríamos lo que podríamos hacer es cambiar las fórmulas (“vectorizarlas”) para que acepten arreglos (“arrays”) de numpy, pero no lo haremos aquí!
```

```python
### Para ello, debemos crear una función para ejecutar la conversión
```

```python
def rapidez_angulo_a_u_v(rapidez, angulo):
    """
    Convertir un vector de velocidad de (rapidez, angulo) a (u, v)
    """
    u = rapidez * math.sin(math.radians(angulo))
    v = rapidez * math.cos(math.radians(angulo))
    return u, v
```

```python
u, v = rapidez_angulo_a_u_v(rapidez1, angulo1)
u, v
```

```python
u, v = rapidez_angulo_a_u_v(rapidez2, angulo2)
u, v
```

```python
## Y si queremos usar la función desde otro de notebook?? Podemos usar Módulos!
```

```python
### Para crear un módulo (“module”) de Python, un archivo separado que contiene la función. Y luego podemos importar este módulo aquí, y desde cualquier notebook u otro módulo
```

```python
%%writefile conversion.py
import math

def rapidez_angulo_a_u_v(rapidez, angulo):
    """
    Convertir un vector de velocidad de (rapidez, angulo) a (u, v)
    """
    u = rapidez * math.sin(math.radians(angulo))
    v = rapidez * math.cos(math.radians(angulo))
    return u, v
```

```python
import conversion
```

```python
u, v = conversion.rapidez_angulo_a_u_v(rapidez1, angulo1)

u, v
```
